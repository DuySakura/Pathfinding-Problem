## 2. Đề xuất project (W3)

### Bài toán
Cho một mê cung được biểu diễn dưới dạng lưới 2 chiều. Nhiệm vụ của bài toán là xác định một đường đi hợp lệ, không đi qua các điểm bị chặn từ điểm xuất phát đến điểm đích sao cho độ dài đường đi là ngắn nhất

### Phương pháp dự kiến
- Bộ dữ liệu được sử dụng: https://movingai.com/benchmarks/grids.html (Dragon Age: Origin, City/street maps, Mazes with variable cooridor width)
- Thử nghiệm các thuật toán: BFS, DFS, DLS, IDS, Greedy best-first, A*, Hill climbing, Genetic algorithm

## 3. Tiến độ giữa kỳ (W8)

### Mô tả các thuật toán tìm đường đã cài đặt:
#### BFS:
- Sử dụng hàng đợi để duyệt theo tầng.
- Tìm đường nhưng không đảm bảo ngắn nhất (do có thêm đường chéo với trọng số $\sqrt{2}$).

#### DFS:
- Sử dụng ngăn xếp để duyệt sâu.
- Tìm đường nhưng không đảm bảo ngắn nhất.

#### DLS:
- Biến thể của DFS với giới hạn độ sâu khi duyệt.
- Tránh thuật toán bị sa lầy quá sâu vào một nhánh, nhưng có thể bỏ lỡ nghiệm nếu ở sâu hơn độ sâu giới hạn.

#### IDS:
- Gọi DLS nhiều lần, độ sâu tối đa tăng sau mỗi lần gọi.
- Kết hợp ưu điểm tiết kiệm bộ nhớ của DFS và khác phục nhược điểm bỏ lỡ nghiệm của DLS.


### Vấn đề gặp phải:
- Hiện tại, nhóm chúng em mới chỉ đánh giá các thuật toán qua các chỉ số cơ bản như tỷ lệ vượt qua testcase, tỉ lệ chênh lệch so với giá trị tối ưu và thời gian thực thi. Chúng em mong nhận được sự gợi ý từ thầy về các chỉ số bổ sung để việc đánh giá toàn diện hơn.
